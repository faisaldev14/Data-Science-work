```python
#LIST METHODS
#• append()- to add item at the end of the list
#• insert() - to add item at any specific position in the list
#• pop()- to delete item using its index position
#• remove() - to delete items using item name
#• copy() - to make a duplicate of the list
#• sort() - to sort the list in a particular order
#• extend() - to combine two list into a single list
#• clear() - to delete all the items from the list but not the list
```

```python
apps = ['google', 'drive','netflix', 'whatsapp','blinkit' ,'snapchat']
print(apps)
```

```python
l1=[1,2,3,4,5,6]
apps.extend(l1)
print(apps)
```

```python
apps.reverse()
print(apps)
```

```python
#- Write a program that creates an empty list and then asks the user to input integers to add to the list. Print the final list.
x = input("Please enter a number: ")
numbers = []
for i in range(int(x)):
  numbers.append(i)
print(numbers)
```

```python
#- Write a program that creates a list of integers and then asks the user to input an integer to check if it exists in the list. Print a message indicating whether the integer is in the list or not.
numbers = [10, 20, 30, 40, 50]

num = int(input("Enter an integer: "))

if num in numbers:
    print("Integer exists in the list.")
else:
    print("Integer does not exist in the list.")
```

```python
#- Write a program that creates a list of strings and then sorts the list alphabetically. Print the sorted list.
names = ["Rahul", "Aman", "Zoya", "Faisal", "Neha"]

names.sort()

print("Sorted list:", names)
```

```python
#- Write a program that creates a list of integers and then finds the maximum value in the list. Print the maximum value.
numbers = [10, 25, 5, 40, 15]

maximum = max(numbers)

print("Maximum value:", maximum)
```

```python
#- Write a program that creates two lists of integers and then finds the common elements between those two lists. Print the common elements.
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]

common = []

for num in list1:
    if num in list2:
        common.append(num)

print("Common elements:", common)
```

```python
#- Write a program that creates a list of strings and then removes all elements that contain a specific substring. Print the final list.
words = ["apple", "banana", "pineapple", "mango", "grapes"]

sub = input("Enter substring to remove: ")

new_list = []

for word in words:
    if sub not in word:
        new_list.append(word)

print("Final list:", new_list)
```

```python
#- Write a program that creates a nested list and then accesses a specific element in the list. Print the accessed element.
numbers = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

print("Accessed element:", numbers[1][2])
```

```python
#- Write a program that creates a list of integers and then removes all duplicates from the list. Print the final list.
numbers = [10, 20, 30, 20, 40, 10, 50]

new_list = []

for num in numbers:
    if num not in new_list:
        new_list.append(num)

print("Final list:", new_list)
```
