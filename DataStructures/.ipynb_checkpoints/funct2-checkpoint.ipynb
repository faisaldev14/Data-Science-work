```python
#1. Function without parameter
def my_func():
    print('hello user')

my_func()  # function calling
```

```python
#1. Function with parameter
def my_name(name):  # parameter
    print(f'my name is {name}')

my_name('faisal')
```

```python
#print even or odd
def even_odd(num):
    if num % 2 == 0:
        print('even number')
    else:
        print('odd number')

even_odd(8)
```

```python
#print area of rectangle using a function
def area(length, breadth):
    result = length * breadth
    print("area of rectangle=", result)

area(10, 5)
```

```python
#count the numbers of vowels in string
def vowel_count(string):
    count = 0
    for char in string:
        if char.lower() in "aeiou":
            count += 1
    return count
print(vowel_count("Hello World"))
```

```python
#check prime no using a function
def prime(num):
    if num < 2:
        return "not prime"

    for i in range(2, num):
        if num % i == 0:
            return "not prime"

    else:
        return "prime"

print(prime(9))
```

```python
#1. Lambda function with map() (Positive/Negative/Zero check)
lst = [2, 12, 10, 4, 5, 0, 4, 6]
num = list(map(lambda x: 'positive' if x > 0 else 'negative' if x < 0 else 'zero', lst))
print(num)
```

```python
#2. Function to check even or odd numbers
def number(num):
    if num % 2 == 0:
        print(f'even number')
    else:
        print(f'odd number')

number(8)
```

```python
#3. Function to calculate the area of a rectangle
def area(length, breadth):
    result = length * breadth
    print(result)

area(11, 5)
```

```python
#4. Function to count vowels in a string
def vowel(string):
    count = 0
    for char in string:
        count += 1
    return count

vowel('abdullah abulbhcmncghcsdhbbdcdsghdjhdscgsgcjvhewvrtwdqw-yhbsbvdfqytddbvdftywehaddtyhfydwaghc')
```

```python
#5. Function to calculate the cube of a number
def cube(n):
    return n ** 3

result = cube(3)
print(result)
```
