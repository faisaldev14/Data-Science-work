```python
#'FUNCTIONS' IS A BLOCK OF CODE DESIGNED TO PERFORM ANY SPECIFIC FUNCTIONALITY REPEATEDLY.
#'TYPES OF FUNCTIONS'
#-user defined FUNCTIONS
   #-parameterized
   #-non parameterized
   #-return
   #-non return
   #-default parameterized
   #-lambda
#-built in FUNCTIONS
   #-print,len,append,sort etc....   
```

```python
#non-parametrized
def greet():  #function definition
    print("hello users!")
    print("welcome to the program")
    print("have some coffee")
    print("enjoy your time here! ")

greet()#function call
    
```

```python
greet()
```

```python
#parametrized function
def sum(a,b):  #parameters
    print(a+b)
sum(2,3)   #arguments 
```

```python
#parametrized function
def sum(a,b,c):  #parameters
    print(a+b+c)
result = sum(2,3,4)
print(result)    
```

```python
def my_country(country = "India"):
    print("I am from " + country)

my_country('Nepal') 
my_country('UK')    
my_country()    
my_country('USA')       
```
