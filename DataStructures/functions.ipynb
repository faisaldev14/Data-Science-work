```python
#'FUNCTIONS' IS A BLOCK OF CODE DESIGNED TO PERFORM ANY SPECIFIC FUNCTIONALITY REPEATEDLY.
#'TYPES OF FUNCTIONS'
#-user defined FUNCTIONS
   #-parameterized
   #-non parameterized
   #-return
   #-non return
   #-default parameterized
   #-lambda
#-built in FUNCTIONS
   #-print,len,append,sort etc....   
```

```python
#non-parametrized
def greet():  #function definition
    print("hello users!")
    print("welcome to the program")
    print("have some coffee")
    print("enjoy your time here! ")

greet()#function call
    
```

```python
#parametrized function
def sum(a,b):  #parameters
    print(a+b)
sum(2,3)   #arguments 
```

```python
#parametrized function
def sum(a,b,c):  #parameters
    print(a+b+c)
result = sum(2,3,4)
print(result)    
```

```python
def average(a,b,c,d,e): #parameters
    print(a+b+c+d+e/5)
average(2,3,4,5,6)    
```

```python
#non return
def sum(a,b,c):
    print(a+b+c)
result= sum(2,3,4)
print(result)    
```

```python
#default parameter
def my_country(country = "India"):
    print("I am from " + country)

my_country('Nepal') 
my_country('UK')    
my_country()    
my_country('USA')       
```

```python
#Take a string from user and write a function to store vowel and consonant character
def vowel_consonant(string):
    vowel_list=[]
    consonant_list=[] 
    for chr in string:
      if chr.lower() in "aeiou":
        vowel_list.append(chr)
      elif chr.isalpha():
        consonant_list.append(chr)
    print('length of vowel list:' ,len(vowel_list))
    print("length of consonant list",len(consonant_list))
    return vowel_list,consonant_list
vowel_consonant ("hello world")
```
