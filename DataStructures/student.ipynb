```python
# Simple program to check if all values in a dictionary are unique

# Example dictionary
my_dict = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 100
}

# Convert dictionary values to a set (sets store only unique items)
values = list(my_dict.values())
unique_values = set(values)

# Compare lengths to check uniqueness
if len(values) == len(unique_values):
    print("All values are unique.")
else:
    print("Values are NOT unique.")
```

```python
# Program to find the difference between two dictionaries

# Example dictionaries
dict1 = {"a": 1, "b": 2, "c": 3}
dict2 = {"b": 2, "c": 4, "d": 5}

# Keys only in first dictionary
only_in_dict1 = {}
for key in dict1:
    if key not in dict2:
        only_in_dict1[key] = dict1[key]

# Keys only in second dictionary
only_in_dict2 = {}
for key in dict2:
    if key not in dict1:
        only_in_dict2[key] = dict2[key]

# Keys in both but with different values
different_values = {}
for key in dict1:
    if key in dict2 and dict1[key] != dict2[key]:
        different_values[key] = (dict1[key], dict2[key])

# Display results
print("Keys only in first dictionary:", only_in_dict1)
print("Keys only in second dictionary:", only_in_dict2)
print("Keys with different values:", different_values)
```
