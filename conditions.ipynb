```python
units = float(input("Enter the number of units consumed: "))

#bill_amount = 0.0


if units <= 100:
    bill_amount = units * 5
elif units <= 300:
      bill_amount = (100 * 5) + ((units - 100) * 7)
else:
    bill_amount = (100 * 5) + (200 * 7) + ((units - 300) * 10)


if bill_amount > 2000:
    surcharge = bill_amount * 0.05
    bill_amount += surcharge
 


print(f"Total Electricity Bill: {bill_amount:.2f}")
```
