- when a column has>40% missing values of the data,then we simple remove that column
- when a column has<40% missing values of data,then we handle those missing values means we replace those missing
values with any number
    - there are two type of column in any dataset-CATAGORICIAL & NUMERICAL
    - when the column is numerical,we always replace the null values with the column mean(average)
    - when the column is categorical is categorical,we always replace the null values with most frequency(mode)values
